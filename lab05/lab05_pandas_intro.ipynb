```python
# 1a
import pandas as pd

pensje = pd.Series({
    'Anna': 5500,
    'Jan': 7200,
    'Ewa': 4800,
    'Marek': 9100,
    'Kasia': 6300
})

print(pensje)
```

```python
#1b
# 1. Kto zarabia najwięcej
print(pensje.idxmax())

# 2. Średnia pensja
print(pensje.mean())

# 3. Osoby powyżej średniej
print(pensje[pensje > pensje.mean()].index)

# 4. Ile osób poniżej 6000
print((pensje < 6000).sum())
```

```python
#1c
# 1. Podwyżka 10%
podwyzka = pensje * 1.10

# 2. Premia 500 zł
premia = pensje + 500

# 3. Nowa pensja (podwyżka + premia)
nowe_pensje = (pensje * 1.10) + 500

# 4. Ranking od najwyższej
ranking = nowe_pensje.sort_values(ascending=False)

print(ranking)
```

## Ćwiczenie 3 

```python
# Dane 
url = 'https://raw.githubusercontent.com/mwaskom/seaborn-data/master/penguins.csv'
penguins = pd.read_csv(url)
```

```python
# 3a
# 1. head(10)
print(penguins.head(10))

# 2. shape
print(penguins.shape)

# 3. info()
penguins.info()
```

```python
# 3b
# 1. describe()
print(penguins.describe())

# 2. describe dla wszystkiego
print(penguins.describe(include='all'))

# 3. Mediana body_mass_g
print(penguins['body_mass_g'].median())
```

```python
# 3c
# 1. Gatunki
print(penguins['species'].value_counts())

# 2. Wyspy
print(penguins['island'].value_counts())

# 3. Płeć
print(penguins['sex'].value_counts())

# 4. Procent Adelie
print(penguins['species'].value_counts(normalize=True))
```

```python
# 3d
# 1. Braki w każdej kolumnie
print(penguins.isna().sum())

# 2. % braków w sex
braki_sex = penguins['sex'].isna().sum()
procent = braki_sex / len(penguins) * 100
print(procent)
```

# Ćwiczenie 4  

```python
# Dane
tips = pd.read_csv('https://raw.githubusercontent.com/mwaskom/seaborn-data/master/tips.csv')
# Dodaj kolumnę z procentem napiwku
tips['tip_pct'] = (tips['tip'] / tips['total_bill'] * 100).round(1)
```

```python
#4
# 1. Średni napiwek (%)
print(tips['tip_pct'].mean())

# 2. Dzień z największą liczbą rachunków
print(tips['day'].value_counts())

# 3. Rachunki od palaczy
print(tips['smoker'].value_counts())

# 4. Describe dla tip + mediana
print(tips['tip'].describe())
print("Mediana napiwku:", tips['tip'].median())

# 5. % rachunków Dinner
print(tips['time'].value_counts(normalize=True))

# 6. Konkretnie Dinner:
print(tips['time'].value_counts(normalize=True)['Dinner'])

# 7. Liczba unikalnych rozmiarów grup
print(tips['size'].nunique())
```
